Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

Create a simple dataset

```python
data = {
    'Date': ['2026-01-01','2026-01-02','2026-01-03','2026-01-04','2026-01-05','2026-01-06'],
    'PM2.5': [80, 95, 70, 110, 90, 75],
    'PM10': [120, 140, 100, 160, 130, 110],
    'CO': [1.2, 1.5, 1.0, 1.8, 1.4, 1.1],
    'AQI': [100, 120, 90, 150, 115, 95]
}

df = pd.DataFrame(data)
df.to_csv('City_Air_Quality.csv', index=False)
```

Import the CSV and explore it

```python
df = pd.read_csv('City_Air_Quality.csv')

print(df.head())
print(df.shape)
print(df.info())
```

Identify relevant variables

```python
df['Date'] = pd.to_datetime(df['Date'])

print(df[['Date', 'PM2.5', 'PM10', 'CO', 'AQI']])
```

Overall AQI Trend

```python
plt.plot(df['Date'], df['AQI'], marker='o')
plt.title('AQI Trend Over Time')
plt.xlabel('Date')
plt.ylabel('AQI')
plt.grid()
plt.show()
```

Pollutant Trends

```python
pollutants = ['PM2.5', 'PM10', 'CO']

for p in pollutants:
    plt.plot(df['Date'], df[p], marker='o', label=p)
    plt.title(p + ' Trend Over Time')
    plt.xlabel('Date')
    plt.ylabel(p)
    plt.legend()
    plt.grid()
    plt.show()
```

AQI Bar Plot

```python
plt.bar(df['Date'].dt.strftime('%d-%m'), df['AQI'])
plt.title('AQI Comparison by Date')
plt.xlabel('Date')
plt.ylabel('AQI')
plt.show()
```

AQI Box Plot

```python
plt.boxplot(df['AQI'])
plt.title('AQI Distribution')
plt.ylabel('AQI')
plt.show()
```

AQI vs Pollutants — Scatter Plots

```python
for p in pollutants:
    plt.scatter(df[p], df['AQI'])
    plt.title('AQI vs ' + p)
    plt.xlabel(p)
    plt.ylabel('AQI')
    plt.grid()
    plt.show()
```

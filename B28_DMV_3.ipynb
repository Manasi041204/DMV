Import Pandas

```python
import pandas as pd
```

Create the Telecom Customer Churn dataset

```python
data = {
    'CustomerID': ['C001', 'C002', 'C003', 'C004', 'C005', 'C006', 'C006'],
    'Gender': ['Male', 'female', 'Male', 'Female', 'male', 'Female', 'Female'],
    'Age': [25, 35, 40, 28, 30, 200, 30],
    'Tenure': [12, 24, 8, 36, 15, 10, 10],
    'MonthlyCharges': [500, 700, 450, 800, None, 600, 600],
    'Contract': ['Monthly', 'Yearly', 'monthly', 'Yearly', 'Monthly', 'monthly', 'monthly'],
    'Churn': ['No', 'Yes', 'No', 'Yes', 'No', 'yes', 'yes']
}

df = pd.DataFrame(data)

df.to_csv('Telecom_Customer_Churn.csv', index=False)

print("Dataset created successfully!")
```

Load and explore the dataset

```python
df = pd.read_csv('Telecom_Customer_Churn.csv')

print("First 5 records:")
display(df.head())

print("Shape:", df.shape)

print("\nDataset Information:")
df.info()
```

Check missing values and duplicates

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

Remove duplicates and fix inconsistencies

```python
# Remove duplicate records
df = df.drop_duplicates()

# Standardize text values
df['Gender'] = df['Gender'].str.capitalize()
df['Contract'] = df['Contract'].str.capitalize()
df['Churn'] = df['Churn'].str.capitalize()

print("Cleaned values:")
display(df)
```

Handle missing values

```python
df['MonthlyCharges'] = df['MonthlyCharges'].fillna(
    df['MonthlyCharges'].mean()
)

print("After handling missing values:")
print(df.isnull().sum())
```

Handle outlier

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df['Age'] >= lower) & (df['Age'] <= upper)]

print("After removing outliers:")
display(df)
```

Feature Engineering

```python
df['TotalCharges'] = df['Tenure'] * df['MonthlyCharges']

print("After Feature Engineering:")
display(df)
```

Final validation

```python
print("Final Dataset:")
display(df)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

print("\nFinal Shape:", df.shape)
```
